<h2> Feature Engineering </h2>

After exploring the data, you can choose to modify features that you would use to predict the performance of the students on your chosen response variable. 

You can also create your own features. For example, perhaps you figured that maybe a state's expenditure per student may affect their overall academic performance so you create a expenditure_per_student feature.

Use this space to modify or create features.

```python
# @@@ 5
# total spending is mostly just a size thing (california vs wyoming) so this is more comparable
df["INSTRUCTION_EXPENDITURE_PER_STUDENT"] = df["INSTRUCTION_EXPENDITURE"] / df["ENROLL_ALL"]
df["SUPPORT_SERVICES_EXPENDITURE_PER_STUDENT"] = df["SUPPORT_SERVICES_EXPENDITURE"] / df["ENROLL_ALL"]

# fill missing enrollment estimates the same way as the tutorial
df["ENROLL_ALL_EST"] = df["ENROLL_ALL_EST"].fillna(df["ENROLL_ALL"])
```

Feature engineering justification: **Bigger states spend more just because they have more kids, so I made instruction spending per student. I did the same for support services. That should be a fairer comparison than using total expenditure.**

<h2> Data Creation </h2>

_Use this space to create train/test data_

```python
from sklearn.model_selection import train_test_split
```

```python
# @@@ 8

# using 4th grade scores + spending per student to predict 8th grade math
X = df[['AVG_MATH_4_SCORE', 'AVG_READING_4_SCORE', 'INSTRUCTION_EXPENDITURE_PER_STUDENT']].dropna()
y = df.loc[X.index]['AVG_MATH_8_SCORE']

# fill leftover missing y values with the median like the tutorial
y.fillna(y.median(), inplace=True)
print(X.shape, y.shape)
```

```python
# @@@ 9 

# 70/30 split, random_state so it doesn't change every time I rerun
X_train, X_test, y_train, y_test = train_test_split(
     X, y, test_size=0.3, random_state=42)
```

<h2> Prediction </h2>

ML Models [Resource](https://medium.com/@vijaya.beeravalli/comparison-of-machine-learning-classification-models-for-credit-card-default-data-c3cf805c9a5a)

```python
# @@@ 10
# import your sklearn class here
from sklearn.linear_model import LinearRegression
```

```python
# @@@ 11
# create your model here
model = LinearRegression()
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

## Evaluation

Choose some metrics to evaluate the performance of your model, some of them are mentioned in the tutorial.

```python
# @@@ 12
# r^2
print(model.score(X_test, y_test))

# mean error
print(np.mean(y_pred - y_test))

# mean absolute error
print(np.mean(np.abs(y_pred - y_test)))

# rmse, this one punishes bigger misses more
print(np.mean((y_pred - y_test)**2)**0.5)

print(model.intercept_)
print(model.coef_)
```
